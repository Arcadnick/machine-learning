### Индексация

В NumPy работает привычная индексация Python, включая использование отрицательных индексов и срезов (slices).

````python
array = np.arange(6)
print(array) # весь массив
print(array[0]) # первый элемент
print(array[-1]) # Последний элемент
print(array[1:-1]) # срез всех элементов массива без первого и последнего
print(array[1:-1:2]) # срез всех элементов массива без первого и последнего с шагом 2
print(array[::-1]) # срез массива в обратном порядке
````

<b>Замечание:</b> Индексы и срезы в многомерных массивах не нужно разделять квадратными скобками, 
т.е. вместо <b>matrix[i][j]</b> нужно использовать <b>matrix[i, j]</b>

<b>Замечание:</b> Срезы в NumPy создают view ("вид"), а не копии данных, поэтому при изменении среза изменится и изначальный массив.

В качестве индексов можно использовать списки

### Расширение размерности

[Broadcasting](https://numpy.org/doc/stable/user/basics.broadcasting.html) позволяет производить операции над массивами разных, но всё-же согласованных размерностей.

![Imgur](https://i.imgur.com/tE3ZCWG.gif)

````python
2 * np.arange(1, 4)
````

Для того, чтобы numpy смог автоматически согласовать размерности, необходимо чтобы размерности либо совпадали либо имели единичный размер.

Прибавить строчку к матрице можно. То есть 1-мерный массив по умолчанию считается вектором-строкой:

````python
a = np.arange(12).reshape(3,4)
print(a.shape)
b = np.arange(4)
print(b.shape)
a + b
````

А вот строчку к столбцам - нет.

````python
a = np.arange(12).reshape(3,4)
print(a.shape)
b = np.arange(3)
print(b.shape)
try:
    a + b
except Exception as e:
    print(e)
````

Для этого есть трюк с фиктивным расширением размерности. Используя структуру `matrix[np.newaxis, :]` или `matrix[None,:]` мы вектор размерности (n) преобразуем к размерности (1, n). Причём перевыделения памяти не происходит, создаётся view с заданной размерностью:

````python
a = np.arange(12).reshape(3,4)
print(a.shape)
b = np.arange(3)
print(b.shape)
print(b[:,np.newaxis].shape)
a + b[:,np.newaxis]
````

Записи `matrix[np.newaxis, :]` или `matrix[None,:]` делают одно и то же, но запись с None короче и универсальнее, в других библиотеках, предоставляющих инструменты для работы с матрицами функционал такой-же.

````python
a = np.arange(12).reshape(3,4)
print(a.shape)
b = np.arange(3)
print(b.shape)
print(b[:,None].shape)
a + b[:,None]
````

Допустим мы хотим получит попарные произведения каждых значений из двух массивов разного размера. То есть хотим получить что-то такое: $c_{ij}=a_i*b_j,\forall i,j$. Можно написать цикл по $i$ и $j$, а можно воспользоваться трюком с расширением размерности:

````python
a = np.arange(1, 5)
b = np.arange(1, 4)
a[:, None] * b[None, :]
````

Библиотека numpy также поддерживает матричные произведения

Сделать матричное произведение можно следующими способами:
- `np.dot(a,b)`
- `a.dot(b)`
- `a @ b`

Все варианты делают одно и то же действие.

````python
a = np.array([[-1,2],[2,1]])
b = np.array([[3,4],[4,5]])

print(f"a={a}")
print(f"b={b}")
print(f"np.dot(a, b)=\n{np.dot(a,b)}")
print(f"a.dot(b)=\n{a.dot(b)}")
print(f"a @ b=\n{a @ b}")
````

Транспонировать матрицу можно с помощью `a.T` или `a.transpose()`. 

`.transpose()` предоставляет больше функциональности, позволяя переставлять размерности, а не только менять порядок на противоположный.

````python
a = np.arange(40).reshape(5,2,4)
print(a.shape)
print(a.T.shape)
print(a.transpose().shape)
print(a.transpose(1, 2, 0).shape)   # перестановка размерностей
````

numpy поддерживает удобные функции для агрегации: max, argmax, sum, mean, prod и другие.

Функции можно вызвать как для конкретного объекта, так и из библиотеки.

````python
print(f"a.max(): {a.max()}")
print(f"a.sum(): {a.sum()}")
print(f"np.sum(a): {np.sum(a)}")
# Это выдаст индекс, если матрицу развернуть в одномерный массив
print(f"a.argmax(): {a.argmax()}") 
# Это выдаст настоящее положение в матрице
print(f"координата в матрице a.argmax(): {np.unravel_index(a.argmax(), a.shape)}") 
````

Для всех операций можно указать параметр `axis` $-$ размерность по которой вычисляется операция. То есть после вычисления суммы указанная размерность `axis` пропадёт:

````python
print(f"a.shape: {a.shape}")
print(f"a.max(axis=1): {a.max(axis=0)}")
print(f"shape: {a.max(axis=0).shape}")
````

Можно передать несколько размерностей:

````python
print(f"a.shape: {a.shape}")
print(f"a.max(axis=(0, 2)): {a.max(axis=(0, 2))}")
print(f"shape: {a.max(axis=(0, 2)).shape}")
````

Также в numpy есть возможность векторизовать функции. То есть сделать их применимыми к целому массиву.

````python
def fun(x):
    return abs(x) * max(x, 5)
````

Просто так применить функцию к массиву не получится:

````python
try:
    fun(np.arange(10))
except Exception as e:
    print(e)
````

А если применить метод `np.vectorise`, то всё заработает:

````python
f = np.vectorize(fun)
f(np.arange(10, dtype=np.float32))
````

Но... Это работает долго. Это всё равно тот же самый цикл из Python, но просто чуть побыстрее.

````python
from time import time
start = time()
f(np.arange(10000000, dtype=np.float32))
print(f"Время работы: {time() - start} сек.")
````

А можно просто написать функцию, но с использованием элементов numpy. Тут вычисления уже идут в хорошо оптимизированном коде библиотеки.

````python
def fun_np(x):
    return np.abs(x) * np.fmax(x, 5) # поэлементный максимум с заданным значением
print("Результат тот же:")
print(fun_np(np.arange(10, dtype=np.float32)))
print("Но быстрее:")
start = time()
fun_np(np.arange(10000000, dtype=np.float32))
print(f"Время работы: {time() - start} сек.")
````

numpy также предоставляет множество операций линейной алгебры. Доступны они в [np.linalg](https://numpy.org/doc/stable/reference/routines.linalg.html).

# Matplotlib

Это библиотека для рисования графиков 

Мы сейчас познакомимся с основными возможностями библиотеки.

В библиотеке есть 2 основных объекта $-$ Рисунок (Figure) и график (Subplot или Axes). 
На каждом Рисунке можно разместить несколько Графиков. Рисунок соответствует непосредственно получаемому изображению, которое можно сохранить в файл или увидеть в ноутбуке или в отдельном окне, если вы запускали скрипт на питоне.

Доступно рисование тепловых карт и линий уровня.

`np.meshgrid(x,y)` полезный метод при рисовании двумерных графиков. Из координат x и y создаёт решётку точек с заданными координатами. Работает так: $x$ понимается как вектор-строка (1, n), a $y$ понимается как вектор-столбец (m, 1) далее размерности обоих векторов расширяются так, чтобы получились две матрицы размера (m, n).


$$x=\begin{pmatrix}x_1\\\vdots\\x_n\end{pmatrix} \;\;y=\begin{pmatrix}y_1\\\vdots\\y_m\end{pmatrix}$$

$$X, Y= np.meshgrid(x,y)$$

$$X = \begin{pmatrix}x_1&\dots&x_n\\\vdots&\ddots&\vdots\\x_1&\dots&x_n\end{pmatrix}\;\;
 Y = \begin{pmatrix}y_1&\dots&y_1\\\vdots&\ddots&\vdots\\y_m&\dots&y_m\end{pmatrix}$$



````python
def fun(x, y):
    return np.exp(-np.square(x) - np.square(y))

x = np.linspace(-2, 2, 50)
y = np.linspace(-2, 2, 100)

x, y = np.meshgrid(x, y)

plt.gcf().set_size_inches(10,6)

ax = plt.subplot(1,3,1)
plt.pcolormesh(x, y, fun(x, y))
plt.title("function")

ax = plt.subplot(1,3,2)
plt.pcolormesh(x, y, x, cmap="Blues")   # Можно менять градиент цветов
plt.title("x coordinate")

ax = plt.subplot(1,3,3)
plt.pcolormesh(x, y, y, cmap="rainbow_r",  # Можно менять градиент цветов
               vmin=0,  # Минимальное значение для назначения цвета
               vmax=1)  # Максимальное значение для назначения цвета
plt.title("y coordinate")
````

## Блок Numpy

### 1. Максимальный элемент среди тех, перед которыми стоит ноль

Идея: сдвигаем массив на один элемент. `X[:-1] == 0` — маска «предыдущий элемент равен нулю», применяем её к `X[1:]`.
Если таких элементов нет, возвращаю `None` (у пустого массива нет максимума).

````python
def max_after_zero(X):
    X = np.asarray(X)
    candidates = X[1:][X[:-1] == 0]
    return candidates.max() if candidates.size else None

x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
print(max_after_zero(x))                       # ожидаем 5 (кандидаты: 3, 0, 5)
print(max_after_zero(np.array([1, 2, 3])))     # нулей нет -> None
print(max_after_zero(np.array([0, -4, 0, -1])))  # отрицательные: -1
RESULTS["max_after_zero"] = int(max_after_zero(x))
````

### 2. Ближайший к числу `value` элемент матрицы

Считаем модуль разности для всей матрицы сразу, `argmin` даёт индекс в «развёрнутой» матрице, по нему и берём элемент через `X.flat`.

````python
def nearest_value(X, value):
    X = np.asarray(X)
    return X.flat[np.abs(X - value).argmin()]

X = np.arange(0, 10).reshape((2, 5))
print(X)
print(nearest_value(X, 3.6))   # 4
print(nearest_value(np.random.normal(0, 1, (4, 4)), 0.0))  # элемент, ближайший к нулю
RESULTS["nearest_value"] = int(nearest_value(X, 3.6))
````

### 3. Применение функции к матрице формы (n, m, k)

Функция работает со *списком* из `p` объектов размерности `k`, т.е. с матрицей `(p, k)`. Матрица `(n, m, k)` — это по сути `n·m` объектов размерности `k`.
Значит, достаточно «расплющить» первые две оси: `reshape(-1, k)` даёт матрицу `(n·m, k)`, функция возвращает вектор длины `n·m`, который возвращаем обратно в форму `(n, m)`.

Замечание: в функции есть слагаемое `np.trace(x @ x.T)` — это сумма квадратов *всех* элементов списка, т.е. общая для всех объектов константа. При таком решении она считается по всем `n·m` объектам, что как раз соответствует трактовке «матрица — это набор объектов». Для проверки сравниваю с наивной реализацией через цикл.

````python
def fun(x):
    # код этой функции изменять нельзя
    return np.sum(x, axis=1) + np.trace(x @ x.T)

data = np.random.normal(0, 5, size=(10, 20, 5))
n, m, k = data.shape

result = fun(data.reshape(-1, k)).reshape(n, m)
print("Форма результата:", result.shape)

# проверка циклом (только для контроля!)
objects = [data[i, j] for i in range(n) for j in range(m)]
reference = fun(np.array(objects)).reshape(n, m)
print("Совпадает с проверкой:", np.allclose(result, reference))
RESULTS["fun_shape"] = list(result.shape)
````

### 4. Попарные евклидовы расстояния

Трюк с расширением размерности: `X[:, None, :]` имеет форму `(n, 1, d)`, а `X[None, :, :]` — `(1, n, d)`. При вычитании срабатывает broadcasting и получаем тензор разностей `(n, n, d)`. Дальше сумма квадратов по последней оси и корень.

Второй вариант экономнее по памяти: $\|a-b\|^2 = \|a\|^2 + \|b\|^2 - 2\langle a,b\rangle$, он не создаёт тензор `(n, n, d)`. Оба сравниваю с `scipy.spatial.distance.cdist`.

````python
def pairwise_distances(X):
    diff = X[:, None, :] - X[None, :, :]
    return np.sqrt(np.square(diff).sum(axis=-1))

def pairwise_distances_gram(X):
    sq = np.square(X).sum(axis=1)
    d2 = sq[:, None] + sq[None, :] - 2 * X @ X.T
    return np.sqrt(np.clip(d2, 0, None))   # clip от маленьких отрицательных из-за округления

P = np.random.normal(0, 1, (6, 3))
D = pairwise_distances(P)
print(np.round(D, 3))
print("Совпадает с cdist:", np.allclose(D, cdist(P, P)))
print("Вариант через матрицу Грама совпадает:", np.allclose(pairwise_distances_gram(P), D))
print("Диагональ нулевая, матрица симметрична:", np.allclose(np.diag(D), 0), np.allclose(D, D.T))
````

## Блок Matplotlib

### 1. Центры масс, срединный перпендикуляр и проекции

Центр масс набора — среднее по точкам. Срединный перпендикуляр к отрезку $c_1c_2$ проходит через середину $M=(c_1+c_2)/2$ с направляющим вектором $u \perp (c_2-c_1)$.
Проекция точки $p$ на прямую: $p' = M + \langle p - M, u\rangle u$ при $\|u\|=1$.

````python
np.random.seed(7)
X1 = np.random.multivariate_normal([4, 4], [[0.1, -0.4], [-0.4, 3]], size=10)
X2 = np.random.multivariate_normal([8, 4], [[0.2, 0.4], [0.4, 1]], size=10)

c1, c2 = X1.mean(axis=0), X2.mean(axis=0)
mid = (c1 + c2) / 2
direction = c2 - c1
u = np.array([-direction[1], direction[0]])
u = u / np.linalg.norm(u)

def project(P):
    return mid + ((P - mid) @ u)[:, None] * u

P1, P2 = project(X1), project(X2)

fig, ax = plt.subplots(figsize=(8, 8))
ax.scatter(*X1.T, label="Набор 1", color="tab:blue")
ax.scatter(*X2.T, label="Набор 2", color="tab:orange")
ax.plot(*np.vstack([c1, c2]).T, "k-", lw=2, label="Отрезок между центрами")
t = np.linspace(-3.5, 3.5, 2)
line = mid + t[:, None] * u
ax.plot(*line.T, "g--", lw=2, label="Срединный перпендикуляр")
for P, Pp, col in [(X1, P1, "tab:blue"), (X2, P2, "tab:orange")]:
    ax.scatter(*Pp.T, marker="x", color=col)
    for p, pp in zip(P, Pp):
        ax.plot([p[0], pp[0]], [p[1], pp[1]], color=col, lw=0.6, alpha=0.6)
for c, name in [(c1, "C1"), (c2, "C2")]:
    ax.scatter(*c, s=150, marker="*", color="red", zorder=5)
    ax.annotate(f"{name} ({c[0]:.2f}; {c[1]:.2f})", xy=c, xytext=(8, 8), textcoords="offset points", fontsize=11)
ax.set_aspect("equal")
ax.set_xlabel("x"); ax.set_ylabel("y")
ax.set_title("Центры масс, срединный перпендикуляр и проекции точек")
ax.legend(loc="best", fontsize=9)
ax.grid(alpha=0.3)
fig.savefig("figures/bisector_projections.png", dpi=150, bbox_inches="tight")
plt.show()

# проверка: проекции лежат на прямой, т.е. перпендикулярны направлению (c2 - c1) относительно середины
print("Проверка:", np.allclose((np.vstack([P1, P2]) - mid) @ direction, 0))
RESULTS["c1"] = c1.round(3).tolist(); RESULTS["c2"] = c2.round(3).tolist()
````

### 2. Сравнение скорости: цикл Python, `np.vectorize`, numpy

Беру функцию с занятия $f(x)=|x|\cdot\max(x,5)$. Размер массива увеличиваю как $10^1 \dots 10^6$, для каждого размера беру минимальное время из нескольких запусков (чтобы меньше влиял шум ОС).

````python
def f_scalar(x):
    return abs(x) * max(x, 5)

def f_loop(arr):
    out = np.empty_like(arr)
    for i in range(len(arr)):
        out[i] = f_scalar(arr[i])
    return out

f_vec = np.vectorize(f_scalar)

def f_numpy(arr):
    return np.abs(arr) * np.fmax(arr, 5)

def bench(func, arr, repeats=3):
    best = np.inf
    for _ in range(repeats):
        start = time.perf_counter()
        func(arr)
        best = min(best, time.perf_counter() - start)
    return best

sizes = np.logspace(1, 6, 11).astype(int)
timings = {"Цикл Python": [], "np.vectorize": [], "numpy": []}
for size in sizes:
    arr = np.random.uniform(-10, 10, size).astype(np.float32)
    assert np.allclose(f_loop(arr[:100]), f_numpy(arr[:100]))
    timings["Цикл Python"].append(bench(f_loop, arr))
    timings["np.vectorize"].append(bench(f_vec, arr))
    timings["numpy"].append(bench(f_numpy, arr))

fig, ax = plt.subplots(figsize=(8, 5))
for name, ts in timings.items():
    ax.plot(sizes, ts, marker="o", label=name)
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("Размер массива, элементов")
ax.set_ylabel("Время работы, с")
ax.set_title("Время поэлементного применения функции")
ax.grid(True, which="both", alpha=0.3)
ax.legend()
fig.savefig("figures/timing.png", dpi=150, bbox_inches="tight")
plt.show()

last = {k: v[-1] for k, v in timings.items()}
print({k: f"{v:.4f} c" for k, v in last.items()})
print(f"numpy быстрее цикла в {last['Цикл Python'] / last['numpy']:.0f} раз, "
      f"быстрее vectorize в {last['np.vectorize'] / last['numpy']:.0f} раз")
RESULTS["timing_1e6"] = last
````

**Вывод.** На лог-лог графике все три зависимости при больших размерах — почти параллельные прямые с наклоном ≈1, то есть время растёт линейно с размером массива. Но константа отличается на порядки: `np.vectorize` лишь немного быстрее явного цикла (внутри всё равно вызывается Python-функция на каждый элемент), а настоящая numpy-реализация работает в скомпилированном коде и быстрее на 2–3 порядка. На маленьких массивах (десятки элементов) разница меньше: там доминируют накладные расходы на вызов функций.
